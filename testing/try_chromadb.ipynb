```python
import chromadb
from chromadb.utils.embedding_functions import OllamaEmbeddingFunction
import json
from pathlib import Path
```

```python
with open(
    r"D:\RAG_Project\storages\outputs\vector_embed\Pedoman PI_embedded.json",
    "r",
    encoding="utf-8",
) as f:
  data = json.load(f)
```

```python
documents = [item["text"] for item in data]
ids = [str(i) for i in range(len(documents))]
```

```python
chroma_client = chromadb.Client()
```

```python
#Hapus collection
chroma_client.delete_collection(name="pdf_sampel")
```

```python
ollama_ef = OllamaEmbeddingFunction(
    url="http://localhost:11434/api/embeddings",
    model_name="bge-m3",
)
```

```python
collection = chroma_client.get_or_create_collection(
    name="pedoman_pi", 
    embedding_function=ollama_ef
)
```

```python
collection.add(ids=ids, documents=documents)
```

```python
query_input = "Bagaimana cara membuat daftar isi?"

results = collection.query(
    query_texts=[query_input],
    n_results=10)
```

```python
print (f"Teks query: {query_input}")
print (f"ID: {results['ids']}")
print (f"Hasil query: {results['documents']}")
print (f"Hasil similarity: {results['distances']}")
print (f"Metadadata: {results['metadatas']}")
```
