```python
class MokaPot:

  # Daftarkan kapasitas DAN isi di blueprint awal
  def __init__(self, kapasitas: str="", isi: int=0):
    self.kapasitas = kapasitas  # Atribut 1
    self.isi = isi  # Atribut 2 (default 0 ml)

  def isi_air(self, ml: int):
    self.isi += ml
```

```python
poci_kopi = MokaPot()

poci_kopi.kapasitas = "6cup"
#poci_kopi.isi = 200  # Menambah 200 ml air
poci_kopi.isi_air(100)  # Menambah 100 ml air

print(f"Kapasitas : {poci_kopi.kapasitas}")  # Output: 6 cup
print(f"Isi awal: {poci_kopi.isi} ml")  # Output: 50 ml
print(f"Isi Akhir : {poci_kopi.isi} ml")  # Output: 150 ml
```
