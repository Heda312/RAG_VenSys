```python
def cek_umur(umur):
    if umur < 17:
        return "Belum cukup umur" # Langsung keluar dari fungsi!
    
    # Kode di bawah ini gak bakal diakses kalau umur < 17
    print("Selamat datang di venue!")
    return "Boleh masuk"

print(cek_umur(18)) # Output: "Belum cukup umur"
```

```python
# SALAH: list 'items' cuma dibuat sekali saat fungsi didefinisikan
def tambah_item(item, items=[]):
    items.append(item)
    return items

print(tambah_item(1))  # [1]
print(tambah_item(2))  # [1, 2] -> Lho, kok 1 nya masih ada?

# BENAR:
def tambah_item_benar(item, items=None):
    if items is None:
        items = []  # Bikin list baru tiap dipanggil
    items.append(item)
    return items

print(tambah_item_benar(1))  # [1]
print(tambah_item_benar(2))  # [2] -> Sekarang 1 nya udah hilang
```
